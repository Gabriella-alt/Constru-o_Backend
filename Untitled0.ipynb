```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from IPython.display import display

# 1. Carregar os dados do dataset iris
from sklearn.datasets import load_iris
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
display(df.head())

# 2. Padronizar os dados
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df)
df_scaled = pd.DataFrame(df_scaled, columns=df.columns)
display(df_scaled.head())

# 3. Aplicar o PCA para reduzir a dimensionalidade em 2 componentes principais
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
df_pca = pca.fit_transform(df_scaled)
df_pca = pd.DataFrame(df_pca, columns=['PC1', 'PC2'])
display(df_pca[:5])

# 4. Configurar o gráfico de dispersão para visualizar os dados após a redução de dimensionalidade com seaborn
sns.scatterplot(x='PC1', y='PC2', data=df_pca, hue=iris.target, palette='Set1')
```

```
# 5. Aplicar o KMeans de forma iterativa para encontrar o número ideal de clusters usando o método do cotovelo
# criamos uma lista vazia para armazenar os valores de inércia (soma das distâncias quadradas dos pontos ao centro do cluster)
inertias = []

# 5.1 - vamos testar o KMeans para diferentes números de clusters (de 1 a 10)
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(df_scaled)
    inertias.append(kmeans.inertia_)

# 5.2 - plotar o gráfico do método do cotovelo
plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), inertias, "bo-", linewidth=2, markersize=8)
plt.title("Método do Cotovelo (Elbow Method) para Determinar o Número K Ideal de Clusters")
plt.xlabel("Número de Clusters (K)")
plt.ylabel("Inércia (Soma das Distâncias Quadradas)")
plt.xticks(range(1, 11)) # Garante que todos os valores de K sejam mostrados no eixo x
plt.grid(True) # Adiciona linhas de grade para melhor visualização
plt.show() # Exibe o gráfico do método do cotovelo


from sklearn.metrics import silhouette_score

fig, axes = plt.subplots(1, 5, figsize=(20, 4), sharex=True, sharey=True)

for k, ax in enumerate(axes, start=1):
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(df_scaled)

    df_plot = df_pca.copy()
    df_plot['Cluster'] = labels
    sns.scatterplot(x='PC1', y='PC2', hue='Cluster', data=df_plot, palette='Set1', ax=ax)

    if k > 1:
        sil = silhouette_score(df_scaled, labels)
        sil_text = f"{sil:.2f}"
    else:
        sil_text = "N/A"

    ax.set_title(f'K = {k} | Inércia = {kmeans.inertia_:.2f} | Silhueta = {sil_text}')
    ax.legend().remove()  # Remove legend for clarity

plt.show()
from sklearn.metrics import silhouette_score

fig, axes = plt.subplots(1, 5, figsize=(20, 4), sharex=True, sharey=True)

for k, ax in enumerate(axes, start=1):
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(df_scaled)

    df_plot = df_pca.copy()
    df_plot['Cluster'] = labels
    sns.scatterplot(x='PC1', y='PC2', hue='Cluster', data=df_plot, palette='Set1', ax=ax)

    if k > 1:
        sil = silhouette_score(df_scaled, labels)
        sil_text = f"{sil:.2f}"
    else:
        sil_text = "N/A"

    ax.set_title(f'K = {k} | Inércia = {kmeans.inertia_:.2f} | Silhueta = {sil_text}')
    ax.legend().remove()  # Remove legend for clarity

plt.show()
```
